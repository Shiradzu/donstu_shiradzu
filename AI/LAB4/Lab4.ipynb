Использование обученной модели

```python
import subprocess
import sys

# Подключаем tf_keras (Keras 2) для совместимости с моделями Teachable Machine
try:
    from tf_keras.models import load_model
except ImportError:
    subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", "tf-keras"])
    from tf_keras.models import load_model

from PIL import Image, ImageOps
import numpy as np

# Отключаем научную нотацию для удобства чтения
np.set_printoptions(suppress=True)

# Загружаем модель через Keras 2 и метки классов
model = load_model("keras_model.h5", compile=False)

with open("labels.txt", "r", encoding="utf-8") as f:
    class_names = f.readlines()

# Подготавливаем массив нужной формы
data = np.ndarray(shape=(1, 224, 224, 3), dtype=np.float32)

# Открываем и обрезаем изображение до 224x224
image = Image.open("10.jpg").convert("RGB")
size = (224, 224)
image = ImageOps.fit(image, size, Image.Resampling.LANCZOS)

# Преобразуем в массив и нормализуем
image_array = np.asarray(image)
normalized_image_array = (image_array.astype(np.float32) / 127.5) - 1
data[0] = normalized_image_array

# Делаем предсказание
prediction = model.predict(data)
index = np.argmax(prediction)
class_name = class_names[index].strip()
confidence_score = prediction[0][index]

print("Class:", class_name[2:])
print("Confidence Score:", confidence_score)
```
